
# 1. Einstieg Algorithmisches Denken 🧮
---
Was bedeutet algorithmisches, compuationelles, de-kompositorisches Denken überhaupt? Ein Vorschlag für eine Definition gibt Martin Paul Eve in *The digital humanities and literary studies*: 

>“What could we include under such rhetoric of ‘digital methods’ and what types of question might such methods answer? There is, undeniably, **a particular type of ‘decompositional’ thinking** that is necessary to use digital approaches. That is to say that computational methods require problems that can be broken into smaller solvable units of addressable and empirically verifiable hypotheses, to which some literary interpretative work is not suited.” (Eve 2022: 6f)

- **Zusammenfassend:** Computergestützte Methoden benötigen Probleme, die in **kleine lösbare Teile** zerlegt werden können. Die Teile müssen empirisch verifizierbar sein. Nicht alle Fragen sind für diesen Prozess geeignet. Das zerlegen von Fragen in kleinere Einheiten kann jedoch geübt werden.

## 1.3 Die Bausteine des algorithmischen, computationellen und de-kompositorischen Denkens erkennen 🧱

Schau Sie sich die Schritte an, die wir für die Stephen-King-Analyse aufgeschrieben haben. Ohne es zu merken, haben Sie bereits einige **Kernprinzipien des Programmierens** (Karsdorp et al. 2021)angewendet:


#### 1. Variablen (*Variables*)
* **Was es bedeutet:** Mit dem Zuweisungsoperator `=` speichert der Computer Werte unter einem Namen, um später damit zu rechnen.
* **In unserem Beispiel:** Die Gesamtzahl der Wörter (`100000`) und die Zahl der Adverbien (`1500`) werden in Variablen abgelegt, um danach die Quote zu berechnen.

#### 2. Listen (*Lists*)
* **Was es bedeutet:** Eine Liste speichert mehrere Werte in einer festen Reihenfolge.
* **In unserem Beispiel:** Alle Wörter eines Buches bilden eine Liste, ebenso die Bücher, die wir analysieren wollen.

#### 3. Bedingungen (*Conditional expressions*)
* **Was es bedeutet:** Mit `if`, `elif` und `else` wird Code nur ausgeführt, wenn eine Regel erfüllt ist.
* **In unserem Beispiel:** **WENN** ein Wort auf *-ly* endet, **DANN** zähle es +1. Ansonsten wird es ignoriert.

#### 4. Funktionen (*Defining functions*)
* **Was es bedeutet:** Ein fertig gedachter Ablauf wird in eine feste „Anleitung“ verpackt, die man immer wieder aufrufen kann.
* **In unserem Beispiel:** `berechne_adverb_quote(buch)` statt jedes Mal alle Teilschritte neu aufzuschreiben.

#### 5. Schleifen (*Looping*)
Schleifen lassen sich auf drei Ebenen verstehen:

* **Als Python-Konstrukt:** Eine Anweisung wird wiederholt, bis eine Bedingung erfüllt ist, z. B. einmal für jedes Element einer Sequenz (`for wort in woerter:`). Statt denselben Befehl 100.000-mal aufzuschreiben, lassen wir den Computer ihn automatisch wiederholen.
* **Als Denkmuster:** Wir erkennen, dass sich ein Problem aus gleichartigen Teilschritten zusammensetzt. Wir **zerlegen** es, bis wir einen Schritt haben, der sich für jedes Element wiederholen lässt. *„Wiederhole das Prüfen und Zählen für jedes einzelne Wort im Buch."*
* **Als Arbeitsprozess:** Wir arbeiten in Durchgängen. Zuerst **testen** wir die Schleife an einem winzigen Teil (ein Kapitel von *Carrie*). Funktioniert sie, **erweitern** wir sie auf den ganzen Datensatz: *„Wiederhole die Analyse für jedes Buch in der Liste."*, also alle 60+ Romane von Stephen King.
